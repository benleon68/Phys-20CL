```python
# Imports
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy import constants
```

```python
# Read data from csv files
db_slit_data = pd.read_csv("/Users/benleon/Physics Projs/Phys 20CL/data/Double_slit_data.csv")
single_slit_data = pd.read_csv("/Users/benleon/Physics Projs/Phys 20CL/data/Single_slit_data.csv")
grating_data = pd.read_csv("/Users/benleon/Physics Projs/Phys 20CL/data/Grating_data.csv")

# Useful Parameters 
y = single_slit_data["Distance from Center (mm)"]  # Distance from the Central Maximum
L = single_slit_data["Distance L (mm)"] # Distance from slit to screen
m = single_slit_data["Fringe Order m"] # Order of fringe m 
lamda_red = 650e-3 # Convert wavelength to microns
```

```python
## Part 1

theta = np.arctan(y / L)
sin_theta = np.sin(theta)

slit_widths = m * lamda_red / sin_theta
avg_a = np.mean(slit_widths)
std_a = np.std(slit_widths, ddof = 1)
std_mean_a = std_a / np.sqrt(len(slit_widths))

print(f"The manufacturer has the slit width = 80 microns, and averaging the calculated slit width using the experimental data we get: a = {avg_a:.2f} \u00B1 {std_mean_a: .2f} microns")
```

```python
## Part 2

y_db = db_slit_data["Distance from Center (mm)"]  # Distance from the Central Maximum
L_db = db_slit_data["Distance L (mm)"] # Distance from slit to screen
m_db = db_slit_data["Fringe Order m"] # Order of fringe m

theta_db = np.arctan(y_db / L_db)
sin_theta_db = np.sin(theta_db)
d = m_db * lamda_red / sin_theta_db
avg_d = np.mean(d)
std_d = np.std(d, ddof = 1)
std_mean_d = std_a / np.sqrt(len(d))

small_ang_check = y_db / L_db
percent_diff = 100 * np.abs(sin_theta_db - small_ang_check) / sin_theta_db
print(np.mean(percent_diff))


print(f"The manufacturer has the distance between slits as 125 microns, and averaging the calculated distance using the experimental data we get: d = {avg_d:.2f} \u00B1 {std_mean_d: .2f} microns")
```

```python
## Part 3

y_grating = grating_data["Distance From Center(mm)"]  # Distance from the Central Maximum
L_grating = grating_data["Distance L (mm)"] # Distance from slit to screen
m_grating = grating_data["Fringe Order m"] # Order of fringe m

theta_grating = np.arctan(y_grating / L_grating)
sin_theta_grating = np.sin(theta_grating)
d_grating = m_grating * lamda_red / (sin_theta_grating) 
N = 1000 / d_grating 
avg_N = np.mean(N)
std_N = np.std(N, ddof = 1)
std_mean_N = std_N / np.sqrt(len(N))


print(f"The manufacturer has the groove density of the diffraction grating as 100 lines per mm, and averaging the calculated distances between slits using the experimental data we get: N = {avg_N:.2f} \u00B1 {std_mean_N: .2f} lines per mm")
```
